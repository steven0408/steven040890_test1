### Build Dataset

```python
import os
import shutil
from tqdm import tqdm
import json

# ---------------------------
# 原始資料夾 & 目標資料夾
# ---------------------------
src_root = "mvtec-ad"       # MVTec 原始資料
dst_root = "dataset_lora"   # 轉換後資料集

normal_dir = os.path.join(dst_root, "normal")
defect_dir = os.path.join(dst_root, "defect")
caption_dir = os.path.join(dst_root, "captions")

os.makedirs(normal_dir, exist_ok=True)
os.makedirs(defect_dir, exist_ok=True)
os.makedirs(caption_dir, exist_ok=True)

meta = {"normal": [], "defect": []}

# ---------------------------
# 遍歷 MVTec 各類別
# ---------------------------
for category in os.listdir(src_root):
    cat_path = os.path.join(src_root, category)
    if not os.path.isdir(cat_path):
        continue

    # ---- 收集正常資料 (train/good) ----
    normal_src = os.path.join(cat_path, "train", "good")
    if os.path.exists(normal_src):
        for i, fname in enumerate(tqdm(os.listdir(normal_src), desc=f"{category} normal")):
            if not fname.lower().endswith((".png", ".jpg", ".jpeg")):
                continue
            src = os.path.join(normal_src, fname)
            dst_fname = f"{category}_normal_{i:04d}.png"
            dst = os.path.join(normal_dir, dst_fname)
            shutil.copy(src, dst)

            # 自動生成 prompt txt
            txt_path = os.path.join(caption_dir, dst_fname.rsplit(".",1)[0]+".txt")
            prompt = f"a clean {category} product photo"
            with open(txt_path, "w") as f:
                f.write(prompt)

            # 加入 meta
            meta["normal"].append({
                "file": f"normal/{dst_fname}",
                "prompt": prompt
            })

    # ---- 收集異常資料 (test/defects) ----
    test_dir = os.path.join(cat_path, "test")
    if os.path.exists(test_dir):
        for defect_type in os.listdir(test_dir):
            defect_path = os.path.join(test_dir, defect_type)
            if not os.path.isdir(defect_path) or defect_type == "good":
                continue
            for i, fname in enumerate(os.listdir(defect_path)):
                if not fname.lower().endswith((".png", ".jpg", ".jpeg")):
                    continue
                src = os.path.join(defect_path, fname)
                dst_fname = f"{category}_{defect_type}_{i:04d}.png"
                dst = os.path.join(defect_dir, dst_fname)
                shutil.copy(src, dst)

                # 自動生成 prompt txt
                txt_path = os.path.join(caption_dir, dst_fname.rsplit(".",1)[0]+".txt")
                prompt = f"a {category} product with visible {defect_type}"
                with open(txt_path, "w") as f:
                    f.write(prompt)

                # 加入 meta
                meta["defect"].append({
                    "file": f"defect/{dst_fname}",
                    "prompt": prompt
                })

# ---------------------------
# 儲存 meta.json
# ---------------------------
with open(os.path.join(dst_root, "meta.json"), "w") as f:
    json.dump(meta, f, indent=2)

print("資料整理完成 ✅")
```

```python
# ============================================================
# style_lora_training_stagewise_full.py
# 三階段式 LoRA + Style + (可選) Defect/Contrastive 訓練流程
# ============================================================

import os
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import CLIPTokenizer
from diffusers import StableDiffusionPipeline, DDPMScheduler
from peft import LoraConfig
import matplotlib.pyplot as plt
from tqdm import tqdm

# LPIPS 或 fallback VGG
use_lpips = False
try:
    import lpips
    loss_fn_lpips = lpips.LPIPS(net="vgg").cuda() if torch.cuda.is_available() else lpips.LPIPS(net="vgg")
    use_lpips = True
    print("Using LPIPS for style loss.")
except Exception as e:
    print("LPIPS not available, fallback to VGG perceptual loss.")
    from torchvision import models
    import torch.nn.functional as F
    vgg = models.vgg16(pretrained=True).features.eval()
    if torch.cuda.is_available():
        vgg = vgg.cuda()
    def vgg_features(x, model=vgg):
        feats = []
        current = x
        for i, layer in enumerate(model):
            current = layer(current)
            if i in {3, 8, 15, 22}:
                feats.append(current)
        return feats

# ---------------------------
# Config
# ---------------------------
MODEL_ID = "runwayml/stable-diffusion-v1-5"
DATASET_DIR = "dataset_lora"
OUTPUT_DIR = "lora_mctev_style_stage"
IMAGE_SIZE = 512
BATCH_SIZE = 2
LR_UNET = 2e-4
LR_VAE = 1e-4
EPOCHS = 5
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
LORA_R = 8
STYLE_START = 0.30
STYLE_MAX = 0.80
STAGE1_EPOCHS = 2
STAGE2_EPOCHS = 1
T_MIN = 200
T_MAX = 700
GRAD_CLIP = 1.0
SAVE_VAE_FINE_TUNE = True
DEBUG_IMAGE_PATH = "debug_samples"
os.makedirs(DEBUG_IMAGE_PATH, exist_ok=True)

# ---------------------------
# Dataset
# ---------------------------
class MVTecDataset(Dataset):
    def __init__(self, root_dir, tokenizer, size=512):
        self.image_dir = os.path.join(root_dir, "normal")
        self.caption_dir = os.path.join(root_dir, "captions")
        self.tokenizer = tokenizer
        self.size = size
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))])

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        img_path = os.path.join(self.image_dir, img_name)
        image = Image.open(img_path).convert("RGB").resize((self.size, self.size))
        image = np.array(image).astype(np.float32)/255.0
        pixel_values = torch.from_numpy(image).permute(2,0,1) * 2 - 1

        caption_path = os.path.join(self.caption_dir, img_name.rsplit(".",1)[0]+".txt")
        if os.path.exists(caption_path):
            with open(caption_path, "r") as f:
                caption = f.read().strip()
        else:
            caption = "a clean product photo"

        inputs = self.tokenizer(
            caption, padding="max_length", truncation=True,
            max_length=self.tokenizer.model_max_length,
            return_tensors="pt"
        )
        return {"pixel_values": pixel_values,
                "input_ids": inputs.input_ids.squeeze(0),
                "attention_mask": inputs.attention_mask.squeeze(0)}

# ---------------------------
# Load model
# ---------------------------
tokenizer = CLIPTokenizer.from_pretrained('openai/clip-vit-large-patch14')
pipe = StableDiffusionPipeline.from_pretrained(MODEL_ID, torch_dtype=torch.float32, safety_checker=None).to(DEVICE)
pipe.scheduler = DDPMScheduler(num_train_timesteps=1000)

dataset = MVTecDataset(DATASET_DIR, tokenizer, IMAGE_SIZE)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

# ---------------------------
# LoRA on UNet
# ---------------------------
lora_config = LoraConfig(
    r=LORA_R, lora_alpha=16,
    target_modules=["to_q","to_k","to_v","to_out.0"],
    lora_dropout=0.05, bias="none"
)
pipe.unet.add_adapter(lora_config)

# ---------------------------
# VAE decoder grad control
# ---------------------------
vae = pipe.vae
def set_requires_grad_for_vae_decoder(vae_model, requires_grad=True):
    for name, param in vae_model.named_parameters():
        if "decoder" in name or "post_quant_conv" in name or "conv_" in name:
            param.requires_grad = requires_grad
        else:
            param.requires_grad = False

# ---------------------------
# Scheduler precompute
# ---------------------------
alphas_cumprod = pipe.scheduler.alphas_cumprod.to(DEVICE)
sqrt_alphas_cumprod = alphas_cumprod.sqrt()
sqrt_one_minus_alphas_cumprod = (1 - alphas_cumprod).sqrt()
def predict_x0_from_eps(z_t, eps, t):
    a_cum = alphas_cumprod[t].view(-1,1,1,1)
    sa = sqrt_alphas_cumprod[t].view(-1,1,1,1)
    som = sqrt_one_minus_alphas_cumprod[t].view(-1,1,1,1)
    return (z_t - som*eps)/sa

# ---------------------------
# Style loss
# ---------------------------
def style_loss_fn(recon, target):
    if use_lpips:
        return loss_fn_lpips(recon, target).mean()
    else:
        recon_v = (recon+1)/2
        target_v = (target+1)/2
        mean = torch.tensor([0.485,0.456,0.406], device=recon.device).view(1,3,1,1)
        std  = torch.tensor([0.229,0.224,0.225], device=recon.device).view(1,3,1,1)
        recon_norm = (recon_v-mean)/std
        target_norm = (target_v-mean)/std
        f_recon = vgg_features(recon_norm)
        f_target = vgg_features(target_norm)
        loss = 0
        for fr,ft in zip(f_recon,f_target):
            loss += F.mse_loss(fr,ft)
        return loss

# ---------------------------
# Training loop (stagewise)
# ---------------------------
pipe.unet.train()
vae.train()
pipe.text_encoder.eval()
loss_history = []
batch_loss_history = []
ema_loss = None

for epoch in range(EPOCHS):
    # Stage control
    if epoch < STAGE1_EPOCHS:
        stage = 1
        set_requires_grad_for_vae_decoder(vae, False)
        print(f"\nEpoch {epoch+1}: STAGE 1 (LoRA-only, style focus)")
    elif epoch < STAGE1_EPOCHS + STAGE2_EPOCHS:
        stage = 2
        set_requires_grad_for_vae_decoder(vae, False)
        print(f"\nEpoch {epoch+1}: STAGE 2 (short noise tuning)")
    else:
        stage = 3
        set_requires_grad_for_vae_decoder(vae, True)
        print(f"\nEpoch {epoch+1}: STAGE 3 (VAE decoder + style)")

    # Refresh optimizer params
    unet_params = [p for p in pipe.unet.parameters() if p.requires_grad]
    vae_params  = [p for p in vae.parameters() if p.requires_grad]
    optimizer = torch.optim.AdamW(
        [{"params": unet_params, "lr": LR_UNET},
         {"params": vae_params, "lr": LR_VAE}],
        betas=(0.9,0.999), weight_decay=0.01
    )

    running_loss = 0
    for step,batch in enumerate(tqdm(dataloader)):
        pixel_values = batch["pixel_values"].to(DEVICE)
        input_ids = batch["input_ids"].to(DEVICE)

        # encode
        with torch.no_grad():
            posterior = vae.encode(pixel_values)
            z = posterior.latent_dist.sample()*0.18215 if hasattr(posterior,'latent_dist') else posterior*0.18215

        # noise
        t = torch.randint(T_MIN, T_MAX, (z.shape[0],), device=DEVICE).long()
        noise = torch.randn_like(z)
        z_noisy = pipe.scheduler.add_noise(z, noise, t)

        # text encode
        with torch.no_grad():
            encoder_hidden_states = pipe.text_encoder(input_ids)[0]

        # UNet
        noise_pred = pipe.unet(z_noisy, t, encoder_hidden_states).sample

        # losses
        loss_noise = torch.nn.functional.mse_loss(noise_pred, noise)
        with torch.enable_grad():
            z0_pred = predict_x0_from_eps(z_noisy, noise_pred, t)
            recon_image = vae.decode(z0_pred/0.18215).sample.clamp(-1,1)
            loss_style = style_loss_fn(recon_image, pixel_values)

        style_weight = min(STYLE_START + (epoch+1)/EPOCHS*(STYLE_MAX-STYLE_START), STYLE_MAX)
        loss = loss_noise + style_weight*loss_style

        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(pipe.unet.parameters(), GRAD_CLIP)
        torch.nn.utils.clip_grad_norm_(vae.parameters(), GRAD_CLIP)
        optimizer.step()

        batch_loss_history.append(loss.item())
        running_loss += loss.item()
        ema_loss = 0.9*ema_loss+0.1*loss.item() if ema_loss else loss.item()

        if step % 10 == 0:
            print(f"  [Epoch {epoch+1} | Step {step}] loss={loss.item():.4f} (noise={loss_noise.item():.4f}, style={loss_style.item():.4f}, w={style_weight:.2f})")

    avg = running_loss/len(dataloader)
    loss_history.append(avg)
    print(f"Epoch {epoch+1}/{EPOCHS} | Avg Loss={avg:.6f}, EMA={ema_loss:.6f}")

    # debug sample
    try:
        with torch.no_grad():
            sample_img_path = os.path.join(dataset.image_dir, dataset.image_files[0])
            sample_img = Image.open(sample_img_path).convert("RGB").resize((IMAGE_SIZE, IMAGE_SIZE))
            prompt = "a product with a small surface defect"
            out = pipe(prompt=prompt, image=sample_img, strength=0.6, guidance_scale=7.5, num_inference_steps=30).images[0]
            out.save(os.path.join(DEBUG_IMAGE_PATH, f"sample_epoch_{epoch+1}.png"))
    except Exception as e:
        print("Debug sample generation failed:", e)

# --- Save models ---
pipe.unet.save_attn_procs(os.path.join(OUTPUT_DIR,"lora_unet"))
if SAVE_VAE_FINE_TUNE:
    vae_state = {k:v.cpu().clone() for k,v in vae.state_dict().items() if "decoder" in k or "post_quant_conv" in k}
    torch.save(vae_state, os.path.join(OUTPUT_DIR,"vae_decoder_finetune.pth"))

# --- Plot loss ---
plt.figure(figsize=(8,5))
plt.plot(batch_loss_history,label="Batch Loss", alpha=0.5)
plt.plot(np.linspace(0,len(batch_loss_history),len(loss_history)),
         loss_history, marker="o", label="Epoch Avg Loss", linewidth=2, color="red")
plt.title("Loss Curve (Batch & Epoch)")
plt.xlabel("Batch (approx.)")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# style_lora_training_stagewise_fixed.py
# 三階段式 LoRA + Style + (可選) Defect/Contrastive 訓練流程（保守版）
# 目的：避免破壞原始分布，同時學習 style 與 defect 特徵

import os
import math
import json
import random
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import CLIPTokenizer
from diffusers import StableDiffusionPipeline, DDPMScheduler
from peft import LoraConfig
import matplotlib.pyplot as plt
from tqdm import tqdm
import torchvision

# ---------------------------
# Config (保守預設，可再調)
# ---------------------------
MODEL_ID = "runwayml/stable-diffusion-v1-5"
DATASET_DIR = "dataset_lora"
OUTPUT_DIR = "lora_mctev_style_stage_fixed"
IMAGE_SIZE = 512
BATCH_SIZE = 2
# ---------- lowered LR to avoid drift ----------
LR_UNET = 5e-5     # originally 2e-4
LR_VAE = 2e-5      # originally 1e-4
EPOCHS = 5
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
LORA_R = 4          # lowered rank
LORA_ALPHA = 16
STYLE_START = 0.20
STYLE_MAX = 0.50
STAGE1_EPOCHS = 2
STAGE2_EPOCHS = 1
# Stage2 t-range narrower to avoid high-noise drift
T_MIN = 200
T_MAX = 700
STAGE2_T_MIN = 300
STAGE2_T_MAX = 500
GRAD_CLIP = 1.0
SAVE_VAE_FINE_TUNE = True
DEBUG_IMAGE_PATH = "debug_samples_fixed"
SAVE_LOG_EVERY = 1    # epochs
os.makedirs(DEBUG_IMAGE_PATH, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)

# style loss frequency and recon weight
STYLE_FREQ = 2       # compute style loss every N steps
RECON_WEIGHT = 0.5   # L1 weight to anchor pixel distribution

# Contrastive/classifier settings (optional)
USE_CONTRASTIVE = True   # stage3 will use contrastive if defect data exists
CONTRASTIVE_WEIGHT = 0.8
CE_WEIGHT = 0.5
TEMPERATURE = 0.1
USE_CLASSIFIER = True

# ---------------------------
# LPIPS or VGG fallback
# ---------------------------
use_lpips = False
try:
    import lpips
    loss_fn_lpips = lpips.LPIPS(net="vgg").cuda() if torch.cuda.is_available() else lpips.LPIPS(net="vgg")
    use_lpips = True
    print("Using LPIPS for style loss.")
except Exception:
    print("LPIPS not available, using VGG features fallback.")
    from torchvision import models
    import torch.nn.functional as F
    vgg = models.vgg16(pretrained=True).features.eval()
    if torch.cuda.is_available():
        vgg = vgg.cuda()
    def vgg_features(x, model=vgg):
        feats = []
        current = x
        for i, layer in enumerate(model):
            current = layer(current)
            if i in {3, 8, 15, 22}:
                feats.append(current)
        return feats

# ---------------------------
# Dataset (normal images + optional captions)
# ---------------------------
class MVTecDataset(Dataset):
    def __init__(self, root_dir, tokenizer, size=512):
        self.image_dir = os.path.join(root_dir, "normal")
        self.caption_dir = os.path.join(root_dir, "captions")
        self.tokenizer = tokenizer
        self.size = size
        self.image_files = sorted([f for f in os.listdir(self.image_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))])

    def __len__(self):
        return len(self.image_files)

    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        img_path = os.path.join(self.image_dir, img_name)
        image = Image.open(img_path).convert("RGB").resize((self.size, self.size))
        image = np.array(image).astype(np.float32)/255.0
        pixel_values = torch.from_numpy(image).permute(2,0,1) * 2 - 1

        caption_path = os.path.join(self.caption_dir, img_name.rsplit(".",1)[0]+".txt")
        if os.path.exists(caption_path):
            with open(caption_path, "r") as f:
                caption = f.read().strip()
        else:
            caption = "a clean product photo"

        inputs = self.tokenizer(caption, padding="max_length", truncation=True,
                                max_length=self.tokenizer.model_max_length,
                                return_tensors="pt")
        return {"pixel_values": pixel_values,
                "input_ids": inputs.input_ids.squeeze(0),
                "attention_mask": inputs.attention_mask.squeeze(0),
                "fname": img_name}

# ---------------------------
# Load pipeline & prepare
# ---------------------------
print("Loading tokenizer and pipeline (this may take a while)...")
tokenizer = CLIPTokenizer.from_pretrained('openai/clip-vit-large-patch14')
pipe = StableDiffusionPipeline.from_pretrained(MODEL_ID, torch_dtype=torch.float32, safety_checker=None).to(DEVICE)
pipe.scheduler = DDPMScheduler(num_train_timesteps=1000)
vae = pipe.vae

dataset = MVTecDataset(DATASET_DIR, tokenizer, IMAGE_SIZE)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True, drop_last=True)

# ---------------------------
# LoRA on UNet (inject) - conservative r
# ---------------------------
lora_config = LoraConfig(
    r=LORA_R,
    lora_alpha=LORA_ALPHA,
    target_modules=["to_q","to_k","to_v","to_out.0"],
    lora_dropout=0.05,
    bias="none"
)
pipe.unet.add_adapter(lora_config)
print("LoRA adapter added (r=%d)." % LORA_R)

# Ensure only LoRA params are trainable at first (protect main UNet)
for name, p in pipe.unet.named_parameters():
    if "lora" in name.lower() or "attn_procs" in name.lower():
        p.requires_grad = True
    else:
        p.requires_grad = False

# ---------------------------
# Defect dataset scan (optional)
# ---------------------------
DEFECT_ROOT = os.path.join(DATASET_DIR, "defect")
defect_types = []
defect_files_by_type = {}
if os.path.exists(DEFECT_ROOT):
    for t in sorted(os.listdir(DEFECT_ROOT)):
        tp = os.path.join(DEFECT_ROOT, t)
        if os.path.isdir(tp):
            files = [os.path.join(tp,f) for f in os.listdir(tp) if f.lower().endswith((".png",".jpg",".jpeg"))]
            if files:
                defect_types.append(t)
                defect_files_by_type[t] = files
num_defect_types = len(defect_types)
if num_defect_types>0:
    print("Detected defect types:", defect_types)
else:
    print("No defect types found; Stage3 contrastive will be unsupervised if enabled.")

def sample_random_defect():
    if num_defect_types==0:
        return None, None
    t = random.choice(defect_types)
    f = random.choice(defect_files_by_type[t])
    img = Image.open(f).convert("RGB").resize((IMAGE_SIZE, IMAGE_SIZE))
    arr = np.array(img).astype(np.float32)/255.0
    pix = torch.from_numpy(arr).permute(2,0,1)*2 - 1
    label = defect_types.index(t)
    return pix, label

# ---------------------------
# Precompute scheduler arrays & helpers
# ---------------------------
alphas_cumprod = pipe.scheduler.alphas_cumprod.to(DEVICE)
sqrt_alphas_cumprod = alphas_cumprod.sqrt()
sqrt_one_minus_alphas_cumprod = (1 - alphas_cumprod).sqrt()

def predict_x0_from_eps(z_t, eps, t):
    a_cum = alphas_cumprod[t].view(-1,1,1,1)
    sa = sqrt_alphas_cumprod[t].view(-1,1,1,1)
    som = sqrt_one_minus_alphas_cumprod[t].view(-1,1,1,1)
    return (z_t - som*eps) / sa

def pool_latent_mean(z):
    # z: [B, C, h, w] -> [B, C]
    return z.mean(dim=[2,3])

def nt_xent_loss(anchor, positive, temperature=0.1):
    a = torch.nn.functional.normalize(anchor, dim=1)
    p = torch.nn.functional.normalize(positive, dim=1)
    logits = torch.matmul(a, p.T) / temperature
    labels = torch.arange(a.shape[0], device=a.device)
    return torch.nn.functional.cross_entropy(logits, labels)

# ---------------------------
# Style loss wrapper
# ---------------------------
def style_loss_fn(recon, target):
    if use_lpips:
        return loss_fn_lpips(recon, target).mean()
    else:
        recon_v = (recon + 1) / 2
        target_v = (target + 1) / 2
        mean = torch.tensor([0.485,0.456,0.406], device=recon.device).view(1,3,1,1)
        std  = torch.tensor([0.229,0.224,0.225], device=recon.device).view(1,3,1,1)
        recon_norm = (recon_v - mean) / std
        target_norm = (target_v - mean) / std
        f_recon = vgg_features(recon_norm)
        f_target = vgg_features(target_norm)
        loss = 0.0
        for fr, ft in zip(f_recon, f_target):
            loss += torch.nn.functional.mse_loss(fr, ft)
        return loss

# ---------------------------
# Training state & loss logs
# ---------------------------
pipe.unet.train()
vae.train()   # we will toggle decoder requires_grad depending on stage
pipe.text_encoder.eval()

loss_history = []
batch_loss_history = []
detailed_log = {"epoch": []}

# classifier lazy-init
classifier = None

# ---------------------------
# Training loop (stagewise with conservative updates)
# ---------------------------
for epoch in range(EPOCHS):
    # determine stage
    if epoch < STAGE1_EPOCHS:
        stage = 1
        set_decoder_grad = False
        print(f"\nEpoch {epoch+1}: STAGE 1 (LoRA-only: style focus)")
    elif epoch < STAGE1_EPOCHS + STAGE2_EPOCHS:
        stage = 2
        set_decoder_grad = False
        print(f"\nEpoch {epoch+1}: STAGE 2 (short noise tuning - narrower t range)")
    else:
        stage = 3
        set_decoder_grad = True
        print(f"\nEpoch {epoch+1}: STAGE 3 (defect learning - allow decoder grads)")

    # configure decoder grads
    for name, param in vae.named_parameters():
        if ("decoder" in name) or ("post_quant_conv" in name) or ("conv_" in name):
            param.requires_grad = bool(set_decoder_grad)
        else:
            param.requires_grad = False

    # build optimizer param groups fresh (safer)
    unet_params = [p for p in pipe.unet.parameters() if p.requires_grad]
    vae_params = [p for p in vae.parameters() if p.requires_grad]
    opt_groups = [{"params": unet_params, "lr": LR_UNET}]
    if len(vae_params) > 0:
        opt_groups.append({"params": vae_params, "lr": LR_VAE})

    # if classifier already created, include it
    if classifier is not None:
        opt_groups.append({"params": classifier.parameters(), "lr": LR_VAE})

    optimizer = torch.optim.AdamW(opt_groups, betas=(0.9,0.999), weight_decay=0.01)

    running_loss = 0.0
    epoch_stats = {"noise":0.0, "recon":0.0, "style":0.0, "contrast":0.0, "ce":0.0}

    for step, batch in enumerate(tqdm(dataloader, desc=f"Epoch {epoch+1}/{EPOCHS}")):
        pixel_values = batch["pixel_values"].to(DEVICE)            # [B,3,H,W]
        input_ids = batch["input_ids"].to(DEVICE)

        # ---- encode to latent (no grad) ----
        with torch.no_grad():
            posterior = vae.encode(pixel_values)
            latent_dist = getattr(posterior, "latent_dist", posterior)
            z = latent_dist.sample() if hasattr(latent_dist, "sample") else latent_dist
            z = z * 0.18215    # scale

        # ---- choose timesteps (narrower for stage2) ----
        if stage == 2:
            t = torch.randint(STAGE2_T_MIN, STAGE2_T_MAX, (z.shape[0],), device=DEVICE).long()
        else:
            t = torch.randint(T_MIN, T_MAX, (z.shape[0],), device=DEVICE).long()

        noise = torch.randn_like(z)
        z_noisy = pipe.scheduler.add_noise(z, noise, t)

        # ---- text encode (no grad) ----
        with torch.no_grad():
            encoder_hidden_states = pipe.text_encoder(input_ids)[0]

        # ---- UNet predict ----
        noise_pred = pipe.unet(z_noisy, t, encoder_hidden_states).sample

        # ---- base noise loss ----
        loss_noise = torch.nn.functional.mse_loss(noise_pred, noise)

        # ---- reconstruct z0_pred and decode (enable grad if decoder trainable) ----
        with torch.enable_grad():
            z0_pred = predict_x0_from_eps(z_noisy, noise_pred, t)
            decode_input = z0_pred / 0.18215
            decoded = vae.decode(decode_input).sample  # usually in [-1,1]
            recon_image = decoded.clamp(-1,1)

        # ---- pixel reconstruction L1 loss (anchor distribution) ----
        loss_recon = torch.nn.functional.l1_loss(recon_image, pixel_values)

        # ---- style loss computed sparsely to avoid over-pulling ----
        if (step % STYLE_FREQ) == 0:
            loss_style = style_loss_fn(recon_image, pixel_values)
        else:
            loss_style = torch.tensor(0.0, device=DEVICE)

        # ---- Stage 3: optionally compute contrastive + classifier losses ----
        loss_contrast = torch.tensor(0.0, device=DEVICE)
        loss_ce = torch.tensor(0.0, device=DEVICE)
        if stage == 3 and USE_CONTRASTIVE:
            # sample one defect (may be None if no defect data)
            defect_pix, defect_label = sample_random_defect()
            if defect_pix is not None:
                defect_pix = defect_pix.to(DEVICE).unsqueeze(0)  # [1,3,H,W]
                with torch.no_grad():
                    post_def = vae.encode(defect_pix)
                    lat_def = (getattr(post_def, "latent_dist", post_def).sample() ) * 0.18215
                # pool
                anchor = pool_latent_mean(z)       # [B, C]
                positive = pool_latent_mean(lat_def)  # [1, C]
                if anchor.shape[0] != positive.shape[0]:
                    positive = positive[0].unsqueeze(0).repeat(anchor.shape[0], 1)

                loss_contrast = nt_xent_loss(anchor, positive, temperature=TEMPERATURE)
                # lazy init classifier if enabled and defect types known
                if USE_CLASSIFIER and num_defect_types > 0:
                    if classifier is None:
                        D = anchor.shape[1]
                        classifier = torch.nn.Linear(D, num_defect_types).to(DEVICE)
                        # rebuild optimizer to include classifier
                        opt_groups.append({"params": classifier.parameters(), "lr": LR_VAE})
                        optimizer = torch.optim.AdamW(opt_groups, betas=(0.9,0.999), weight_decay=0.01)
                        print("Initialized classifier with input dim", D)
                    logits = classifier(anchor)
                    target_labels = torch.full((anchor.shape[0],), defect_label, dtype=torch.long, device=DEVICE)
                    loss_ce = torch.nn.functional.cross_entropy(logits, target_labels)

        # ---- combine losses (noise primary) ----
        frac = (epoch + 1) / EPOCHS
        style_weight = min(STYLE_START + frac * (STYLE_MAX - STYLE_START), STYLE_MAX)
        loss = loss_noise + RECON_WEIGHT * loss_recon + style_weight * loss_style
        if stage == 3 and USE_CONTRASTIVE:
            loss = loss + CONTRASTIVE_WEIGHT * loss_contrast + CE_WEIGHT * loss_ce

        # ---- backward & step ----
        optimizer.zero_grad()
        loss.backward()
        torch.nn.utils.clip_grad_norm_(pipe.unet.parameters(), GRAD_CLIP)
        if len(vae_params)>0:
            torch.nn.utils.clip_grad_norm_(vae.parameters(), GRAD_CLIP)
        if classifier is not None:
            torch.nn.utils.clip_grad_norm_(classifier.parameters(), GRAD_CLIP)
        optimizer.step()

        # ---- logging ----
        batch_loss_history.append(loss.item())
        running_loss += loss.item()
        epoch_stats["noise"] += float(loss_noise.item())
        epoch_stats["recon"] += float(loss_recon.item())
        epoch_stats["style"] += float(loss_style.item())
        epoch_stats["contrast"] += float(loss_contrast.item())
        epoch_stats["ce"] += float(loss_ce.item())

        if step % 10 == 0:
            print(f"  [E{epoch+1} S{step}] loss={loss.item():.4f} (noise={loss_noise.item():.4f}, recon={loss_recon.item():.4f}, style={loss_style.item():.4f}, w={style_weight:.3f})")

    # epoch averages
    num_steps = len(dataloader)
    epoch_avg = running_loss / num_steps
    loss_history.append(epoch_avg)
    for k in epoch_stats:
        epoch_stats[k] /= max(1, num_steps)
    detailed_log["epoch"].append({
        "epoch": epoch+1, "avg_loss": epoch_avg, "stats": epoch_stats, "stage": stage
    })
    print(f"Epoch {epoch+1} done | Avg loss {epoch_avg:.6f} | noise {epoch_stats['noise']:.6f} recon {epoch_stats['recon']:.6f} style {epoch_stats['style']:.6f}")

    # save debug images: original, reconstruction, sample output from pipe for visual check
    try:
        # save recon vs orig for first batch
        with torch.no_grad():
            sample_batch = next(iter(DataLoader(dataset, batch_size=min(4,len(dataset)), shuffle=False)))
            orig = sample_batch["pixel_values"][:4].to(DEVICE)
            # encode+decode
            post = vae.encode(orig)
            lat = getattr(post, "latent_dist", post).sample() * 0.18215
            dec = vae.decode(lat / 0.18215).sample.clamp(-1,1)
            torchvision.utils.save_image((orig+1)/2, os.path.join(DEBUG_IMAGE_PATH, f"orig_epoch{epoch+1}.png"))
            torchvision.utils.save_image((dec+1)/2, os.path.join(DEBUG_IMAGE_PATH, f"recon_epoch{epoch+1}.png"))

            # sample pipeline inference using first image (img2img)
            sample_img = (orig[0].cpu().permute(1,2,0).numpy()*255).astype(np.uint8)
            pil = Image.fromarray(sample_img)
            out = pipe(prompt="a product with a small surface defect", image=pil, strength=0.6, guidance_scale=7.5, num_inference_steps=30).images[0]
            out.save(os.path.join(DEBUG_IMAGE_PATH, f"sample_epoch{epoch+1}.png"))
    except Exception as e:
        print("Debug image save failed:", e)

    # checkpoint saves
    if (epoch+1) % SAVE_LOG_EVERY == 0:
        # save LoRA/attn procs
        try:
            pipe.unet.save_attn_procs(os.path.join(OUTPUT_DIR, f"lora_unet_epoch{epoch+1}"))
        except Exception as e:
            print("Warning: saving attn_procs failed:", e)
        if SAVE_VAE_FINE_TUNE and set_decoder_grad:
            try:
                vae_state = {k:v.cpu().clone() for k,v in vae.state_dict().items() if ("decoder" in k) or ("post_quant_conv" in k)}
                torch.save(vae_state, os.path.join(OUTPUT_DIR, f"vae_decoder_finetune_epoch{epoch+1}.pth"))
            except Exception as e:
                print("Warning: saving vae state failed:", e)

# end epochs

# final saves & plots
with open(os.path.join(OUTPUT_DIR, "training_log.json"), "w") as f:
    json.dump(detailed_log, f, indent=2)

torch.save({"loss_history": loss_history, "batch_loss_history": batch_loss_history}, os.path.join(OUTPUT_DIR, "losses.pt"))

# plot losses
plt.figure(figsize=(8,5))
plt.plot(batch_loss_history, label="Batch Loss", alpha=0.5)
plt.plot(np.linspace(0, len(batch_loss_history), len(loss_history)), loss_history, marker="o", label="Epoch Avg Loss", linewidth=2, color="red")
plt.title("Loss Curve (Batch & Epoch)")
plt.xlabel("Batch (approx.)")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "loss_curve.png"))
plt.show()

# save final LoRA+vae
try:
    pipe.unet.save_attn_procs(os.path.join(OUTPUT_DIR, "lora_unet_final"))
except Exception as e:
    print("Final save attn_procs failed:", e)
if SAVE_VAE_FINE_TUNE:
    try:
        vae_state = {k:v.cpu().clone() for k,v in vae.state_dict().items() if ("decoder" in k) or ("post_quant_conv" in k)}
        torch.save(vae_state, os.path.join(OUTPUT_DIR, "vae_decoder_finetune_final.pth"))
    except Exception as e:
        print("Saving final VAE state failed:", e)

print("Training finished. Outputs saved to", OUTPUT_DIR)
```

```python
# inference_mctev_pairs.py
# 針對你現有 training outputs 做 per-category「正常 vs 生成異常」對照
# 左：正常圖片，右：生成的異常圖片

import os
import random
import torch
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
from diffusers import StableDiffusionPipeline

# ----------------- Config (可調) -----------------
MODEL_ID = "runwayml/stable-diffusion-v1-5"
# 與你的 training script 相同的 OUTPUT_DIR
OUTPUT_DIR = "lora_mctev_style"   # <- 若不同請改
DATASET_ROOTS = ["mvtec-ad", "mvtec", "mvtc", "dataset_lora"]  # 依序嘗試
IMAGE_SIZE = 512
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# inference hyperparams
STRENGTH = 0.2
GUIDANCE_SCALE = 7.5
NUM_STEPS = 100
import random

SEED = random.randint(0, 2**32 - 1)

SAVE_DIR = os.path.join(OUTPUT_DIR, "inference_pairs")
os.makedirs(SAVE_DIR, exist_ok=True)

# optional prompt map for nicer prompts per category (擴充用)
PROMPT_MAP = {
    "leather": "a leather surface with scratches and scuffs",
    "bottle": "a bottle with cracks and dirt stains",
    "cable": "a cable with fraying and exposed wires",
    "tile": "a ceramic tile with cracks and chips",
    "metal_nut": "a metal nut with rust and deformation",
    "toothbrush": "a toothbrush with bristle damage and dirt",
    "hazelnut": "a hazelnut with shell cracks and surface holes",
    "screw": "a screw with bent head and corrosion",
    "zipper": "a zipper with missing teeth and tears",
    # add more category-specific prompts if you want
}

# ----------------- helper: find categories & one normal image per category -----------------
def find_category_normals():
    # Try various dataset roots; return dict: {category: normal_image_path}
    found = {}
    # 1) check for classic MVTec layout: <root>/<category>/train/good/*.png
    for root in DATASET_ROOTS:
        if not os.path.exists(root):
            continue
        # if root has category subfolders (mvtec-ad style)
        entries = [d for d in os.listdir(root) if os.path.isdir(os.path.join(root, d))]
        # heuristics: if subfolders contain train/good, assume mvtec structure
        mvtec_like = False
        for d in entries[:10]:
            if os.path.isdir(os.path.join(root, d, "train", "good")):
                mvtec_like = True
                break
        if mvtec_like:
            for cat in entries:
                train_good = os.path.join(root, cat, "train", "good")
                if os.path.isdir(train_good):
                    imgs = [f for f in os.listdir(train_good) if f.lower().endswith((".png", ".jpg", ".jpeg"))]
                    if imgs:
                        found[cat] = os.path.join(train_good, imgs[0])
            if found:
                return found

        # 2) dataset_lora style: root/normal/*.png where filenames may include category prefix
        normal_dir = os.path.join(root, "normal")
        if os.path.isdir(normal_dir):
            imgs = [f for f in os.listdir(normal_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))]
            # try to group by prefix before first underscore
            groups = {}
            for f in imgs:
                prefix = f.split("_")[0]
                groups.setdefault(prefix, []).append(os.path.join(normal_dir, f))
            # pick one per prefix
            for prefix, files in groups.items():
                if prefix not in found:
                    found[prefix] = files[0]
            if found:
                return found

        # 3) fallback: if root itself contains image files, treat root as single category
        imgs = [f for f in os.listdir(root) if f.lower().endswith((".png", ".jpg", ".jpeg"))]
        if imgs:
            cat = os.path.basename(os.path.abspath(root))
            found[cat] = os.path.join(root, imgs[0])
            return found

    return found

# ----------------- Load pipeline and LoRA -----------------
print("Loading pipeline...")
pipe = StableDiffusionPipeline.from_pretrained(MODEL_ID, torch_dtype=torch.float32, safety_checker=None).to(DEVICE)

# load LoRA attn procs if present
lora_dir = os.path.join(OUTPUT_DIR, "lora_unet")
if os.path.exists(lora_dir):
    try:
        pipe.unet.load_attn_procs(lora_dir)
        print("Loaded LoRA from", lora_dir)
    except Exception as e:
        print("Warning: failed to load LoRA from", lora_dir, ":", e)
else:
    # maybe saved per-epoch or final name differs; try file search
    candidates = [p for p in os.listdir(OUTPUT_DIR) if p.startswith("lora_unet")]
    if candidates:
        try:
            pipe.unet.load_attn_procs(os.path.join(OUTPUT_DIR, candidates[0]))
            print("Loaded LoRA from", os.path.join(OUTPUT_DIR, candidates[0]))
        except Exception as e:
            print("Warning loading LoRA candidate:", e)
    else:
        print("No lora_unet found under", OUTPUT_DIR, "- inference will use base UNet (not ideal).")

# try load fine-tuned VAE if exists
vae_path = os.path.join(OUTPUT_DIR, "vae_decoder_finetune_final.pth")
if not os.path.exists(vae_path):
    # maybe saved per-epoch
    candidates = [f for f in os.listdir(OUTPUT_DIR) if f.startswith("vae_decoder_finetune")]
    if candidates:
        vae_path = os.path.join(OUTPUT_DIR, candidates[-1])
if os.path.exists(vae_path):
    try:
        state = torch.load(vae_path, map_location=DEVICE)
        # In training we saved partial dict (only decoder keys). load with strict=False
        pipe.vae.load_state_dict(state, strict=False)
        print("Loaded fine-tuned VAE from", vae_path)
    except Exception as e:
        print("Warning: loading VAE failed:", e)
else:
    print("No fine-tuned VAE found; using base VAE.")

# ----------------- Prepare categories -----------------
cat_normals = find_category_normals()
if not cat_normals:
    raise RuntimeError("找不到任何 category 或 normal 圖片。請確認資料結構或把 MVTec 資料放在 mvtec-ad/ 或 dataset_lora/。")

print("Found categories:", list(cat_normals.keys()))

# ----------------- Inference loop -----------------
def generate_anomaly_for_image(pipe, pil_img, prompt, strength=STRENGTH, guidance=GUIDANCE_SCALE, steps=NUM_STEPS, seed=SEED):
    generator = torch.manual_seed(seed)
    out = pipe(prompt=prompt, image=pil_img, strength=strength, guidance_scale=guidance, num_inference_steps=steps, generator=generator).images[0]
    return out

for cat, normal_path in sorted(cat_normals.items()):
    try:
        print(f"Processing {cat} -> {normal_path}")
        pil = Image.open(normal_path).convert("RGB").resize((IMAGE_SIZE, IMAGE_SIZE))
    except Exception as e:
        print("Failed to open", normal_path, e)
        continue

    prompt = PROMPT_MAP.get(cat.lower(), f"a {cat} with a small surface defect or anomaly")

    try:
        gen = generate_anomaly_for_image(pipe, pil, prompt, strength=STRENGTH, guidance=GUIDANCE_SCALE, steps=NUM_STEPS, seed=SEED)
    except Exception as e:
        print("Generation failed for", cat, ":", e)
        try:
            gen = generate_anomaly_for_image(pipe, pil, prompt, strength=0.5, guidance=6.5, steps=20, seed=SEED)
            print("Fallback generation succeeded for", cat)
        except Exception as e2:
            print("Fallback also failed:", e2)
            continue

    # show side-by-side in a Matplotlib window
    fig, axes = plt.subplots(1,2, figsize=(10,5))
    axes[0].imshow(pil); axes[0].set_title(f"{cat} - Normal"); axes[0].axis("off")
    axes[1].imshow(gen); axes[1].set_title(f"{cat} - Generated Anomaly"); axes[1].axis("off")
    plt.tight_layout()
    plt.show()   # 互動式環境會彈出視窗或在 Notebook 中顯示
    plt.close(fig)

print("All done. Displayed images for each category.")
```
